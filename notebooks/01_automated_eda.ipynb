```python
import pandas as pd
from ydata_profiling import ProfileReport

# 1. Load the data from the root data/ folder
df = pd.read_csv('data/telco_churn.csv')

# 2. Generate the profile report
profile = ProfileReport(df, title="Customer Churn EDA Report", explorative=True)

# 3. Save the HTML dashboard directly into the notebooks folder
profile.to_file("notebooks/eda_report.html")

# 4. (Bonus) Render the dashboard directly inside this notebook cell!
profile.to_notebook_iframe()
```
